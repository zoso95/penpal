# Rainbow Noise Rings

Concentric rings warped by noise, each ring in a different color layer.

```python
import penpal
from penpal import noise
import numpy as np

d = penpal.Drawing(10, 10, center=True)
pw = penpal.pen_width(0.3)

# Rainbow color cycle
colors = ['#E91E3A', '#FF6F00', '#FFD600', '#43A047', '#1E88E5', '#7B1FA2', '#E91E8C']
color_names = ['red', 'orange', 'yellow', 'green', 'blue', 'purple', 'pink']
for name, color in zip(color_names, colors):
    d.layer(name, color=color, linewidth=pw)

# Build one polar mesh with many rings, warp with noise
mesh = (penpal.Mesh.polar(center=(0, 0), inner_r=1.0, outer_r=4.5, rings=70, spokes=80)
        .warp_noise(amplitude=0.6, frequency=0.5, seed=42, pin_edges=True))

# Extract each ring as a separate smooth closed curve, assign to color layer
n_spokes, n_rings = mesh.shape
for j in range(n_rings):
    # Close the ring
    ring_x = np.concatenate([mesh.x[:, j], mesh.x[:1, j]])
    ring_y = np.concatenate([mesh.y[:, j], mesh.y[:1, j]])
    ring = np.column_stack([ring_x, ring_y])
    # Assign to color layer by cycling
    layer_name = color_names[j % len(colors)]
    d.layer(layer_name).add(penpal.Paths([ring]))

d
```

```python
d.save('output/rainbow_rings.svg')
d.save_layers('output/rainbow_rings')
```
