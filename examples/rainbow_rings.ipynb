# Rainbow Noise Rings

Concentric rings warped by noise, each ring in a different color layer.

```python
import penpal
import numpy as np

d = penpal.Drawing(10, 10, center=True)
pw = penpal.pen_width(0.3)

# Rainbow color cycle
colors = ['#E91E3A', '#FF6F00', '#FFD600', '#43A047', '#1E88E5', '#7B1FA2', '#E91E8C']
color_names = ['red', 'orange', 'yellow', 'green', 'blue', 'purple', 'pink']
for name, color in zip(color_names, colors):
    d.layer(name, color=color, linewidth=pw)

# Radial noise: displaces rings in/out coherently — rings don't cross
mesh = (penpal.Mesh.polar(center=(0, 0), inner_r=1.2, outer_r=4.0, rings=70, spokes=300)
        .warp_radial_noise(amplitude=0.8, freq_angular=2.5, freq_radial=0.2, seed=42))

for j, ring in enumerate(mesh.rings(smooth=False)):
    d.layer(color_names[j % len(colors)]).add(penpal.Paths([ring]))

d
```

```python
d.save('output/rainbow_rings.svg')
d.save_layers('output/rainbow_rings')
```
