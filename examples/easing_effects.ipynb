# Easing Functions & Effects

Visualize easing curves and use them to modulate density, spacing, and warp fields. 20+ functions from linear to elastic bounce.

```python
import penpal
from penpal.effects import easing
import numpy as np

pw = penpal.pen_width(0.25)
```

## Easing Function Gallery
Plot each easing function as a curve.

```python
d = penpal.Drawing(12, 10)
pw2 = penpal.pen_width(0.2)

funcs = [
    ('linear', easing.linear),
    ('ease_in', easing.ease_in),
    ('ease_out', easing.ease_out),
    ('ease_in_out', easing.ease_in_out),
    ('ease_in_cubic', easing.ease_in_cubic),
    ('ease_out_cubic', easing.ease_out_cubic),
    ('smoothstep', easing.smoothstep),
    ('smootherstep', easing.smootherstep),
    ('ease_in_expo', easing.ease_in_expo),
    ('ease_out_circ', easing.ease_out_circ),
    ('ease_out_elastic', easing.ease_out_elastic),
    ('ease_out_bounce', easing.ease_out_bounce),
]

colors = ['#264653', '#2a9d8f', '#e9c46a', '#e76f51',
          '#e63946', '#457b9d', '#6a0572', '#f1a208',
          '#0f3460', '#16c79a', '#533483', '#1a1a2e']

t = np.linspace(0, 1, 200)

for i, (name, func) in enumerate(funcs):
    col = i % 4
    row = i // 4
    cx = 1.5 + col * 3
    cy = 1.5 + row * 3
    
    y = func(t)
    curve = np.column_stack([t * 2 + cx - 1, y * 2 + cy - 1])
    d.layer(name, color=colors[i], linewidth=pw2).add(penpal.Paths([curve]))
    
    # Box outline
    box = np.array([[cx-1, cy-1], [cx+1, cy-1], [cx+1, cy+1], [cx-1, cy+1], [cx-1, cy-1]])
    d.layer(f'box_{i}', color='#ddd', linewidth=penpal.pen_width(0.15)).add(penpal.Paths([box]))

d
```

## Easing-Modulated Line Spacing
Use easing to vary the vertical spacing of horizontal lines across the page.

```python
d = penpal.Drawing(10, 6)

n_lines = 40
t = np.linspace(0, 1, n_lines)
y_positions = easing.ease_in_expo(t) * 5 + 0.5

lines = []
for y in y_positions:
    lines.append(np.array([[0.5, y], [9.5, y]]))

d.layer('spaced', color='#264653', linewidth=pw).add(penpal.Paths(lines))
d
```

## Pulse and Wave Functions
Repeating patterns: sawtooth, triangle wave, and pulse.

```python
d = penpal.Drawing(10, 8)

t = np.linspace(0, 3, 500)

waves = [
    ('sawtooth', easing.sawtooth(t, period=1.0), '#e63946'),
    ('triangle', easing.triangle_wave(t, period=1.0), '#2a9d8f'),
    ('pulse', easing.pulse(t % 1, center=0.5, width=0.15), '#e9c46a'),
]

for i, (name, y, color) in enumerate(waves):
    curve = np.column_stack([t * 2.5 + 0.5, y * 1.5 + 1 + i * 2.5])
    d.layer(name, color=color, linewidth=pw).add(penpal.Paths([curve]))

d
```

## Remap with Easing
Transform values from one range to another through an easing curve.

```python
d = penpal.Drawing(10, 6)

# Circle radii modulated by easing
n = 30
angles = np.linspace(0, 2*np.pi, n, endpoint=False)
t = np.linspace(0, 1, n)
radii = easing.remap(t, out_min=0.3, out_max=2.5,
                     easing=easing.ease_out_elastic)

lines = []
theta = np.linspace(0, 2*np.pi, 30)
for i, (a, r) in enumerate(zip(angles, radii)):
    cx = 5 + 3 * np.cos(a)
    cy = 3 + 2 * np.sin(a)
    circle = np.column_stack([cx + abs(r)*0.3*np.cos(theta),
                               cy + abs(r)*0.3*np.sin(theta)])
    lines.append(circle)

d.layer('eased_circles', color='#6a0572', linewidth=pw).add(penpal.Paths(lines))
d
```
