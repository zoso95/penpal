# Voronoi Hatched — multi-pen layers

```python
import penpal
from penpal import sampling, shading
import numpy as np

d = penpal.Drawing(8, 10)
pw = penpal.pen_width(0.3)

# Poisson disk → Voronoi
pts = sampling.poisson_disk(8, 10, min_dist=0.55, seed=42)
regions = sampling.voronoi(pts, bounds=(0, 0, 8, 10))

# 3 pen colors
colors = ['#E91E8C', '#00BCD4', '#5C2D91']
names = ['pink', 'cyan', 'purple']
for name, color in zip(names, colors):
    d.layer(name, color=color, linewidth=pw)

rng = np.random.default_rng(42)
for i, region in enumerate(regions):
    name = names[i % 3]
    angle = rng.uniform(0, 180)
    density = rng.integers(15, 30)
    d.layer(name).add(shading.shade_polygon(region, density=density, angle=angle))
    d.layer(name).add(penpal.Paths([region]))

d
```

```python
d = penpal.Drawing(8, 10)
pw = penpal.pen_width(0.3)

# Poisson disk → Voronoi
pts = sampling.poisson_disk(8, 10, min_dist=0.55, seed=42)
regions = sampling.voronoi(pts, bounds=(0, 0, 8, 10))

# 3 pen colors
colors = ['#E91E8C', '#00BCD4', '#5C2D91']
names = ['pink', 'cyan', 'purple']
for name, color in zip(names, colors):
    d.layer(name, color=color, linewidth=pw)

rng = np.random.default_rng(42)
for i, region in enumerate(regions):
    name = names[i % 3]
    angle = rng.uniform(0, 180)
    density = rng.integers(15, 30)

    err = rng.uniform(-5, 5)
    
    d.layer(name).add(shading.shade_polygon(region, density=density, angle=angle))
    d.layer(name).add(shading.shade_polygon(region, density=density, angle=angle+err))
    d.layer(name).add(penpal.Paths([region]))

d
```

```python
d.save('output/voronoi_hatched.svg')
d.save_layers('output/voronoi_hatched')
```
