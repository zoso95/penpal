# Wallpaper Groups

Periodic tilings using crystallographic symmetries. Define a lattice cell, choose a symmetry pattern, draw in the fundamental domain, and tile across the plane.

```python
import penpal
from penpal.symmetry import WallpaperGroup
from penpal import shading
import numpy as np

pw = penpal.pen_width(0.3)
```

## Lattice setup

`WallpaperGroup(origin, b1, b2)` classifies the lattice as square, rectangle, or rhombus.

```python
wg = WallpaperGroup(origin=(0, 0), b1=(0, 1), b2=(1, 0))
print(wg)
print(f'Valid patterns: {wg.valid_patterns}')
```

## *442 — hatched triangles with 4-fold + mirror symmetry

```python
d = penpal.Drawing(8, 10)
wg = WallpaperGroup(origin=(0, 0), b1=(0, 1), b2=(1, 0))
domain = wg.fundamental_domain('*442')
domain_closed = np.vstack([domain, domain[:1]])
art = shading.shade_polygon(domain_closed, density=20, angle=30)
tiled = wg.generate(art, '*442', nx=8, ny=10, bounds=(0, 0, 8, 10))
d.layer('wp', color='black', linewidth=pw).add(tiled)
d
```

## 442 — pinwheel with 4-fold rotation (no mirrors)

```python
d = penpal.Drawing(8, 10)
wg = WallpaperGroup(origin=(0, 0), b1=(0, 1), b2=(1, 0))
domain = wg.fundamental_domain('442')
domain_closed = np.vstack([domain, domain[:1]])
art = shading.shade_polygon(domain_closed, density=15, angle=60)
tiled = wg.generate(art, '442', nx=8, ny=10, bounds=(0, 0, 8, 10))
d.layer('wp', color='#1E88E5', linewidth=pw).add(tiled)
d
```

## *2222 — cross-hatched with two mirror axes

```python
d = penpal.Drawing(8, 10)
wg = WallpaperGroup(origin=(0, 0), b1=(0, 1), b2=(1, 0))
domain = wg.fundamental_domain('*2222')
domain_closed = np.vstack([domain, domain[:1]])
art1 = shading.shade_polygon(domain_closed, density=12, angle=30)
art2 = shading.shade_polygon(domain_closed, density=12, angle=-30)
tiled = wg.generate(art1 + art2, '*2222', nx=8, ny=10, bounds=(0, 0, 8, 10))
d.layer('wp', color='#5C2D91', linewidth=pw).add(tiled)
d
```

## 4*2 — most complex square group

```python
d = penpal.Drawing(8, 10)
wg = WallpaperGroup(origin=(0, 0), b1=(0, 1), b2=(1, 0))
domain = wg.fundamental_domain('4*2')
domain_closed = np.vstack([domain, domain[:1]])
art = shading.shade_polygon(domain_closed, density=6, angle=15)
tiled = wg.generate(art, '4*2', nx=8, ny=10, bounds=(0, 0, 8, 10))
d.layer('wp', color='#FF6F00', linewidth=pw).add(tiled)
d
```

## Recursive — small 442 cells tiled into larger *2222

```python
d = penpal.Drawing(8, 10)
wg_small = WallpaperGroup(origin=(0, 0), b1=(0, 1), b2=(1, 0))
domain_s = wg_small.fundamental_domain('442')
domain_s_closed = np.vstack([domain_s, domain_s[:1]])
art_small = shading.shade_polygon(domain_s_closed, density=8, angle=45)
cell_small = wg_small.complete(art_small, '442')

# Tile the small pattern into a 2x2 block
level1 = wg_small.tile(cell_small, nx=2, ny=2)

# Use that as content for a larger wallpaper group
wg_big = WallpaperGroup(origin=(0, 0), b1=(0, 2), b2=(2, 0))
tiled = wg_big.generate(level1, '*2222', nx=4, ny=5, bounds=(0, 0, 8, 10))
d.layer('wp', color='black', linewidth=pw).add(tiled)
d
```
