# Polar Patterns & Ribbons

Cosine-eased interpolation between boundary curves, concentric ribbon fills, polar function plots, and polar grids.

```python
import penpal
from penpal.gen import polar, curves
import numpy as np

pw = penpal.pen_width(0.25)
```

## Ribbon Pair
Two sinusoidal boundaries with smooth interpolated fill lines between them.

```python
d = penpal.Drawing(10, 6)

p = polar.ribbon_pair(n_points=300, x_range=(0, 10),
                      width=2.0, amplitude=0.8,
                      frequency=3.0, n_fills=20, seed=42)
p = p.translate(0, 2)

d.layer('ribbon', color='#2a9d8f', linewidth=pw).add(p)
d
```

## Custom Ribbon — Arc to Line
Fill between an arbitrary arc and a straight line.

```python
d = penpal.Drawing(8, 8)

t = np.linspace(0, np.pi, 100)
arc = np.column_stack([3 * np.cos(t) + 4, 3 * np.sin(t) + 2])
line = np.column_stack([np.linspace(1, 7, 100), np.full(100, 6.5)])

p = polar.ribbon(arc, line, n_fills=25, easing='cosine')
d.layer('fill', color='#e63946', linewidth=pw).add(p)
d
```

## Concentric Ribbons
Circular ribbon bands — clean or noisy.

```python
d = penpal.Drawing(8, 8)

# Clean concentric ribbons
p = polar.concentric_ribbons(n_ribbons=4, n_fills=8,
                              inner_r=0.5, outer_r=3.5,
                              noise_amplitude=0.0)
p = p.translate(4, 4)

d.layer('clean', color='#264653', linewidth=pw).add(p)
d
```

```python
d = penpal.Drawing(8, 8)

# Noisy organic concentric ribbons
p = polar.concentric_ribbons(n_ribbons=5, n_fills=10,
                              inner_r=0.3, outer_r=3.5,
                              noise_amplitude=0.15, seed=42)
p = p.translate(4, 4)

d.layer('noisy', color='#6a0572', linewidth=pw).add(p)
d
```

## Polar Functions
Plot r = f(theta) curves — roses, limacons, spirals.

```python
d = penpal.Drawing(12, 8)

funcs = [
    (lambda t: 2 + np.cos(3 * t), '#e63946', 'rose-like'),
    (lambda t: 1 + 2 * np.cos(t), '#2a9d8f', 'limacon'),
    (lambda t: np.abs(np.sin(2 * t)) * 2.5, '#e9c46a', 'four-petal'),
]

for i, (func, color, name) in enumerate(funcs):
    p = polar.polar_function(func, n_points=500)
    cx = 2 + i * 4
    p = p.translate(cx, 4)
    d.layer(name, color=color, linewidth=pw).add(p)

d
```

## Polar Grid
Classic polar coordinate grid — rings plus radial spokes.

```python
d = penpal.Drawing(8, 8)

p = polar.polar_grid(n_rings=8, n_spokes=16,
                     inner_r=0.3, outer_r=3.5)
p = p.translate(4, 4)

d.layer('grid', color='#457b9d', linewidth=pw).add(p)
d
```
