# Flow Fields

```python
import penpal
from penpal.gen import flow
import numpy as np

pw = penpal.pen_width(0.3)
```

## Visualize a field — direction indicators at grid points

```python
d = penpal.Drawing(8, 8, center=True)
field = flow.simplex_field(frequency=0.6, scale=1.2, seed=42)

d.layer('field', color='#999999', linewidth=pw).add(
    flow.show_field(field, -4, -4, 4, 4, nx=25, ny=25)
)
d
```

## Field + streamlines overlaid

```python
d = penpal.Drawing(8, 8, center=True)
field = flow.simplex_field(frequency=0.6, scale=1.2, seed=42)

# Show field direction
d.layer('field', color='#CCCCCC', linewidth=pw).add(
    flow.show_field(field, -4, -4, 4, 4, nx=25, ny=25)
)

# Trace streamlines over it
seeds = flow.seeds_line(-4, 0, -3.5, 0, n=60)
d.layer('flow', color='black', linewidth=pw).add(
    flow.trace(field, seeds, steps=600, step_size=0.04, momentum=0.95, bounds=(-4, -4, 4, 4))
)
d
```

## Multiple ribbon bundles

```python
d = penpal.Drawing(10, 8)
field = flow.fractal_field(frequency=0.4, scale=1.5, octaves=5, seed=17)

for y0 in [2, 4, 6]:
    seeds = flow.seeds_line(0, y0, 0.3, y0, n=60)
    d.layer('flow', color='black', linewidth=pw).add(
        flow.trace(field, seeds, steps=600, step_size=0.04, momentum=0.93, bounds=(0, 0, 10, 8))
    )
d
```

## Curl field — smooth non-crossing flows

```python
d = penpal.Drawing(8, 8, center=True)
field = flow.curl_field(frequency=0.5, seed=42)
seeds = flow.seeds_grid(-3.5, -3.5, 3.5, 3.5, nx=30, ny=30)

d.layer('flow', color='#E91E8C', linewidth=pw).add(
    flow.trace(field, seeds, steps=400, step_size=0.03, momentum=0.8, bounds=(-4, -4, 4, 4))
)
d
```

## Spiral field

```python
d = penpal.Drawing(8, 8, center=True)
field = flow.compose_fields(
    flow.spiral_field(center=(0, 0), tightness=0.15),
    flow.simplex_field(frequency=0.6, scale=0.3, seed=7),
)
seeds = flow.seeds_circle(0, 0, 0.5, n=80)

d.layer('flow', color='#5C2D91', linewidth=pw).add(
    flow.trace(field, seeds, steps=800, step_size=0.03, momentum=0.9, bounds=(-4, -4, 4, 4))
)
d
```

## Domain-warped field — turbulent ribbons

```python
d = penpal.Drawing(10, 6)
field = flow.domain_warp_field(
    flow.simplex_field(frequency=0.7, scale=1.0, seed=5),
    warp_amplitude=2.5, warp_frequency=0.3, seed=42,
)
seeds = flow.seeds_line(0, 3, 0.2, 3, n=100)

d.layer('flow', color='black', linewidth=pw).add(
    flow.trace(field, seeds, steps=700, step_size=0.04, momentum=0.96, bounds=(0, 0, 10, 6))
)
d
```

## Bidirectional tracing

```python
d = penpal.Drawing(8, 8, center=True)
field = flow.simplex_field(frequency=0.5, scale=1.0, seed=99)
seeds = flow.seeds_random(-2, -2, 2, 2, n=150, seed=42)

d.layer('flow', color='black', linewidth=pw).add(
    flow.trace_bidirectional(field, seeds, steps=300, step_size=0.04, momentum=0.9, bounds=(-4, -4, 4, 4))
)
d
```

## Custom field (lambda)

```python
d = penpal.Drawing(8, 8, center=True)

# Vortex pair
def vortex_pair(x, y):
    a1 = np.arctan2(y - 1, x) + np.pi / 2  # CW vortex at (0, 1)
    a2 = np.arctan2(y + 1, x) - np.pi / 2  # CCW vortex at (0, -1)
    # Blend by inverse distance
    r1 = max(np.sqrt(x**2 + (y-1)**2), 0.1)
    r2 = max(np.sqrt(x**2 + (y+1)**2), 0.1)
    w1, w2 = 1/r1, 1/r2
    sx = w1*np.cos(a1) + w2*np.cos(a2)
    sy = w1*np.sin(a1) + w2*np.sin(a2)
    return np.arctan2(sy, sx)

seeds = flow.seeds_grid(-3.5, -3.5, 3.5, 3.5, nx=25, ny=25)
d.layer('flow', color='#00BCD4', linewidth=pw).add(
    flow.trace(vortex_pair, seeds, steps=300, step_size=0.03, momentum=0.5, bounds=(-4, -4, 4, 4))
)
d
```
