# Smooth Noise-Warped Grid

```python
import penpal

d = penpal.Drawing(8, 10)
pw = penpal.pen_width(0.3)

mesh = (penpal.Mesh.rect(0, 0, 8, 10, rows=60, cols=48)
        .warp_noise(amplitude=0.4, frequency=0.6, seed=7))

d.layer('grid', color='black', linewidth=pw).add(mesh.to_paths(smooth=True, points_per_line=120))
d.layer('grid').clip_rect(0, 0, 8, 10)
d
```

```python
d.save('output/noise_grid.svg')
```
