# Error Diffusion Dithering

Classic dithering algorithms: Floyd-Steinberg, Stucki, Jarvis-Judice-Ninke, and Atkinson. Quantize images to dots or line segments for plotter output.

```python
import penpal
from penpal.cv import dither, image
import numpy as np

pw = penpal.pen_width(0.25)
```

## Test Image
A synthetic gradient with some structure.

```python
h, w = 120, 160
Y, X = np.mgrid[0:h, 0:w]
# Horizontal gradient with circular feature
img = X / w * 255
cx, cy = w*0.6, h*0.4
r = np.sqrt((X - cx)**2 + (Y - cy)**2)
img = np.where(r < 30, img * 0.3, img)
img = np.clip(img, 0, 255).astype(np.float64)
print(f'Image: {img.shape}')
```

## Floyd-Steinberg
The classic 4-neighbor error diffusion.

```python
d = penpal.Drawing(8, 6)

p = dither.floyd_steinberg(img, n_levels=2, sigma=0.5,
                            dot_radius=0.4)
p = image.map_to_drawing(p, img.shape, d)

d.layer('fs', color='#1a1a2e', linewidth=pw).add(p)
d
```

## Atkinson
Only diffuses 3/4 of the error — higher contrast, more white space. The classic Mac look.

```python
d = penpal.Drawing(8, 6)

p = dither.atkinson(img, n_levels=2, sigma=0.5,
                     dot_radius=0.4)
p = image.map_to_drawing(p, img.shape, d)

d.layer('atk', color='#264653', linewidth=pw).add(p)
d
```

## Stucki vs Jarvis-Judice-Ninke
Wider kernels for smoother gradients.

```python
d = penpal.Drawing(8, 6)

p = dither.stucki(img, n_levels=2, sigma=0.5, dot_radius=0.4)
p = image.map_to_drawing(p, img.shape, d)
d.layer('stucki', color='#2a9d8f', linewidth=pw).add(p)
d
```

```python
d = penpal.Drawing(8, 6)

p = dither.jarvis_judice_ninke(img, n_levels=2, sigma=0.5,
                                dot_radius=0.4)
p = image.map_to_drawing(p, img.shape, d)
d.layer('jjn', color='#e76f51', linewidth=pw).add(p)
d
```

## Dither to Lines
Instead of dots, render as horizontal line segments — more efficient for plotters.

```python
d = penpal.Drawing(8, 6)

p = dither.dither_to_lines(img, n_levels=2, sigma=0.5,
                            kernel='floyd_steinberg', row_skip=2)
p = image.map_to_drawing(p, img.shape, d)

d.layer('lines', color='#0f3460', linewidth=pw).add(p)
d
```

## Multi-Level Dithering
More than 2 levels creates intermediate gray tones with different dot sizes.

```python
d = penpal.Drawing(8, 6)

p = dither.floyd_steinberg(img, n_levels=4, sigma=1.0,
                            dot_radius=0.3)
p = image.map_to_drawing(p, img.shape, d)

d.layer('multi', color='#533483', linewidth=pw).add(p)
d
```
