# Line Envelopes & String Art

Straight lines whose tangent envelope traces curves — hyperbolas, parabolas, cardioids. Classic string art patterns.

```python
import penpal
from penpal.gen import envelopes, curves
import numpy as np

pw = penpal.pen_width(0.25)
```

## Hyperbolic Diamond
Four-quadrant 1/x line envelopes.

```python
d = penpal.Drawing(8, 8)

p = envelopes.diamond(n_lines=50, size=3.5)
p = p.translate(4, 4)

d.layer('diamond', color='#1d3557', linewidth=pw).add(p)
d
```

## Cardioid Envelope
Connect point i to point 2i around a circle — the envelope is a cardioid.

```python
d = penpal.Drawing(8, 8)

# Cardioid (multiplier=2)
p = envelopes.cardioid_envelope(n_lines=100, radius=3.0, multiplier=2)
p = p.translate(4, 4)
d.layer('cardioid', color='#e63946', linewidth=pw).add(p)

# Draw the base circle
c = curves.circle(center=(4, 4), radius=3.0, num_points=200)
d.layer('circle', color='#a8dadc', linewidth=pw).add(c)

d
```

## Nephroid (multiplier=3)

```python
d = penpal.Drawing(8, 8)

p = envelopes.cardioid_envelope(n_lines=120, radius=3.0, multiplier=3)
p = p.translate(4, 4)
d.layer('nephroid', color='#457b9d', linewidth=pw).add(p)

c = curves.circle(center=(4, 4), radius=3.0, num_points=200)
d.layer('circle', color='#a8dadc', linewidth=pw).add(c)
d
```

## Parabolic Envelope
Lines between two edges create a parabolic tangent envelope.

```python
d = penpal.Drawing(8, 8)

# Four corners — center each pattern at its position
corners = [(2, 2, '#264653'), (6, 2, '#2a9d8f'),
           (2, 6, '#e9c46a'), (6, 6, '#e76f51')]

for dx, dy, color in corners:
    p = envelopes.parabolic_envelope(n_lines=40, size=6, angle=90)
    b = p.bounds()
    # Center the pattern on its target position
    cx = (b[0] + b[2]) / 2
    cy = (b[1] + b[3]) / 2
    p = p.translate(-cx, -cy).scale(0.4, 0.4).translate(dx, dy)
    d.layer(f'para_{dx}_{dy}', color=color, linewidth=pw).add(p)

d
```

## String Art — Custom Curves
Connect evenly-spaced points on one curve to reversed points on another.

```python
d = penpal.Drawing(8, 8)

# Arc + line
t = np.linspace(0, np.pi, 60)
arc = np.column_stack([3 * np.cos(t) + 4, 3 * np.sin(t) + 1])
line = np.column_stack([np.linspace(1, 7, 60), np.full(60, 7.0)])

p = envelopes.string_art(n_lines=60, points_a=arc, points_b=line)
d.layer('strings', color='#6a0572', linewidth=pw).add(p)

# Draw the boundary curves
d.layer('bounds', color='#c0c0c0', linewidth=pw).add(penpal.Paths([arc, line]))
d
```

## Multiplier Sweep
Cardioid family: multiplier 2 through 7.

```python
d = penpal.Drawing(12, 8)
colors = ['#e63946', '#f1a208', '#2a9d8f', '#264653', '#6a0572', '#e76f51']

for i, mult in enumerate(range(2, 8)):
    col = i % 3
    row = i // 3
    cx = 2 + col * 4
    cy = 2 + row * 4
    
    p = envelopes.cardioid_envelope(n_lines=150, radius=1.5, multiplier=mult)
    p = p.translate(cx, cy)
    d.layer(f'mult_{mult}', color=colors[i], linewidth=pw).add(p)
    
    c = curves.circle(center=(cx, cy), radius=1.5, num_points=100)
    d.layer(f'ring_{mult}', color='#d0d0d0', linewidth=penpal.pen_width(0.15)).add(c)

d
```
